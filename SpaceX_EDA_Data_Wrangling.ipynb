```python
import pandas as pd
import numpy as np
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
df.head(10)
df.isnull().sum()
df.dtypes
```

## Task 1 - Calculate number of launches per site

```python
df['LaunchSite'].value_counts()
```

## Task 2 - Calculate number of launches per orbit

```python
df['Orbit'].value_counts()
```

## Task 3 - Calculate number and occurrence of mission outcome per orbit

```python
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

## Task 4 - Create landing outcome label - 1 = success, 0 = failure

```python
bad_outcomes = set(landing_outcomes.keys()[[1,3,5,6,7]])
print(bad_outcomes)
landing_class = []
for outcome in df['Outcome']:
    if outcome in bad_outcomes:
        landing_class.append(0)
    else:
        landing_class.append(1)
df['Class'] = landing_class
df[['Outcome','Class']].head(20)
df['Class'].mean()
df.to_csv('dataset_part_2.csv', index=False)
```
