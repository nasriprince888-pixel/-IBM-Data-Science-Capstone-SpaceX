```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
df.head()
print(df.shape)
```

## Task 1 - FlightNumber vs PayloadMass

```python
sns.catplot(y='PayloadMass', x='FlightNumber', hue='Class', data=df, aspect=5)
plt.xlabel('Flight Number', fontsize=20)
plt.ylabel('Payload Mass (kg)', fontsize=20)
plt.show()
```

## Task 2 - FlightNumber vs LaunchSite

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=5)
plt.xlabel('Flight Number', fontsize=20)
plt.ylabel('Launch Site', fontsize=20)
plt.show()
```

## Task 3 - Payload vs LaunchSite

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=5)
plt.xlabel('Payload Mass (kg)', fontsize=20)
plt.ylabel('Launch Site', fontsize=20)
plt.show()
```

## Task 4 - Success rate per Orbit

```python
sns.barplot(y='Orbit', x='Class', data=df)
plt.xlabel('Success Rate', fontsize=20)
plt.ylabel('Orbit', fontsize=20)
plt.show()
print(df.groupby('Orbit')['Class'].mean())
```

## Task 5 - FlightNumber vs Orbit

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=5)
plt.show()
```

## Task 6 - Payload vs Orbit

```python
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=5)
plt.show()
```

## Task 7 - Success rate per Year

```python
df['Date'] = pd.to_datetime(df['Date'])
year = df['Date'].dt.year
sns.lineplot(x=year, y=df['Class'])
plt.xlabel('Year')
plt.ylabel('Success Rate')
plt.show()
print(df.groupby(year)['Class'].mean())
```

## Conclusion: Success rate increased over years. ES-L1, GEO, HEO, SSO have highest success. VAFB-SLC-4E has no heavy payload launches.
